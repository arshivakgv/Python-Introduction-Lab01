### IV. print() Function

The print() function prints the specified message to the standard output device (screen), or to the text stream file.

````python
# Example
# A newline is printed at the end by default
x = 123
y = True
message = 'Python is fun'

print(x)
print(y)
print(message)
print('Great!')
````

````python
# print without newline at the end
x = 123
y = True
message = 'Python is fun'

print(x, end="")
print(y, end="")
print(message, end="")
print('Great!', end="")
````

````python
# print with * symbol(s) at the end
x = 123
y = True
message = 'Python is fun'

print(x, end="*")
print(y, end="**")
print(message, end="***")
print('Great!', end="****")
````

````python
# Use comma (,) to separate the string and variables
# the default separator is a space
print(x, y, message, 'Great!')
````

````python
# Use comma (,) to separate the string and variables
# set the separator to '|'
print(x, y, message, 'Great!', sep="|")
````

````python
# Use the + operator to concatenate strings and the variables
name = 'Ada'
age = 18
print("My name is " + name + ". I am " + str(age) + " years old.") 

# Note: use str() to convert numeric to str
````

````python
# Use the string formatting with the help of %
name = 'Ada'
age = 18
print("My name is %s. I am %d years old." %(name, age))

# Note:
# %d for integer value
# %f for float value
# %s for string value
````

````python
# Use the string formatting with the help of %
# print float with specified number of decimal places
result = 10/3
print("The result is %f" %(result))
print("The result is %.1f" %(result)) # 1 d.p.
print("The result is %.2f" %(result)) # 2 d.p.
print("The result is %.3f" %(result)) # 3 d.p.
````

````python
# Use the string formatting with the help of {}
name = 'Ada'
age = 18
print("My name is {}. I am {} years old.".format(name, age))
````

````python
# Use the string formatting with the help of {}
# accessing variables by position
name1 = 'Ada'
name2 = 'Bob'
print("{0} is a girl. {1} is a boy.".format(name1, name2))

# Note:
# position 0: name1
# position 1: name2
````

````python
# Use the string formatting with the help of {}
# accessing variables by position
name1 = 'Ada'
name2 = 'Bob'
print("{1} is a girl. {0} is a boy.".format(name1, name2))

# Note:
# position 0: name1
# position 1: name2
````

````python
# Use the f-string in the print statement
name = 'Ada'
age = 18
print(f"My name is {name}. I am {age} years old.")
````

### VI. if statement

Very often when we write code, we want to perform different actions for different decisions. if statement is used to decide whether a certain statement or block of statements will be executed or not. 

#### 1. if

Syntax:

```
if condition:
    Statement(s) to execute if condition is True
```


````python
# Check wheter a number is positive
num = 10
if num > 0:
    print(f'The number is {num}')
    print('positive')
````

````python
# Check wheter a number is positive
# the condition is False, nothing is printed
num = -5
if num > 0: 
    print(f'The number is {num}')
    print('positive')
````

````python
# Check wheter a number is positive
# Does it print anything? why?
num = -5
if num > 0: 
    print(f'The number is {num}')
print('positive')
````

Note: Python uses indentation to identify a block. 

#### 2. if...else

Syntax:

```
if condition:
    Statement(s) to execute if condition is True
else:
    Statement(s) to execute if condition is False

```


````python
# Check wheter a number is positive
num = -5

print(f'The number is {num}')
if num > 0:    
    print('positive')
else:
    print('not positive')
````

#### 3. if...elif...else

Syntax:

```
if condition1:
    Statement(s) to execute if condition1 is True
    
elif condition2:
    Statement(s) to execute if condition2 is True
.
.
.
elif conditionN:
    Statement(s) to execute if conditionN is True
    
else:
    Statement(s) to execute if all conditions are False

```


````python
# Check wheter a number is positive
num = -5

print(f'The number is {num}')
if num > 0:    
    print('positive')
elif num < 0:
    print('negative')
else:
    print('zero')
````

````python
# Check whether a number is common mupliple of 2 and 3

num = 9

if num % 2 == 0 and num % 3 == 0:    # use 'and' to join the condition together
    print(f'{num} is a common mupliple of 2 and 3')
else:
    print(f'{num} is not a common mupliple of 2 and 3')
````

Example: Write a program that ask the user for weight in kilograms and height in meters and then displays the weight class based on BMI in the below table.

Note: $ BMI = weight / height^2 $


| BMI | Weight class |
| --------- | --------- | 
| below 18.5 | Underweight |
| 18.5 - 24.9 | Normal range |
| 25.0 -29.9 | Overweight |
| >= 30 | Obese |

````python
weight = input("Enter your weight(kg): ")
height = input("Enter your height(m): ")
bmi = float(weight) / (float(height)**2)   

print(f'The BMI is {bmi}')
if bmi < 18.5:
    print('Underweight')
elif bmi < 25:
    print('Normal range')
elif bmi < 30:
    print('Overweight')
else:
    print('Obese')
````

1. Write a program to read two integers from user and then print the minimum between the two integers.

Sample output:
```
Enter the first number:  15
Enter the second number:  20

```

````python
firstNumber = int(input("Enter the first number: "))
secondNumber = int(input("Enter the second number: "))
if firstNumber < secondNumber:
    print(firstNumber,"is minimum")
elif firstNumber == secondNumber: 
    print(firstNumber,"and",secondNumber, "are equal")
else:
    print(secondNumber,"is minimum")
````

4. Write a program to read in a year and check whether the year is leap year or not.

Hint: A leap year is exactly divisible by 4 except for century years (years ending with 00). The century year is a leap year only if it is perfectly divisible by 400.

Sample output 1:
```
Enter a year: 2012
2012 is a leap year
```

Sample output 2:
```
Enter a year: 1900
1900 is not a leap year
```

````python
year = int(input("Enter a year: "))
if year % 4 == 0 and (year % 100 != 0 or year % 400 == 0):
    print(year,"is a leap year")
else:
    print(year,"is not a leap year")
````

5. Write a program to input socres of three tests, calcuate the average score and print the grade according to the following table.


| Average score | Grade |
